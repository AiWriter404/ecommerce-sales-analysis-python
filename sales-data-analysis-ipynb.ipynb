# E-Commerce Sales Analysis

This project analyzes e-commerce sales performance across different product categories, regions, and customer segments.

Tools Used:
- Python
- Pandas
- Matplotlib

```python
df['revenue'].sum()
```

```python
category_revenue = (
    df.groupby('product_category')['revenue']
    .sum()
    .sort_values(ascending=False)
)

category_revenue
```

```python
region_revenue = (
    df.groupby('region')['revenue']
    .sum()
    .sort_values(ascending=False)
)

region_revenue
```

```python
payment_revenue = (
    df.groupby('payment_method')['revenue']
    .sum()
    .sort_values(ascending=False)
)

payment_revenue
```

```python
df['customer_rating'].mean()
```

```python
import matplotlib.pyplot as plt

category_revenue.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.ylabel('Revenue')

plt.show()
```

```python
region_revenue.plot(kind='bar')

plt.title('Revenue by Region')

plt.show()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
monthly_sales = (
    df.groupby(df['order_date'].dt.to_period('M'))['revenue']
    .sum()
)

monthly_sales.plot(figsize=(12,5))

plt.title("Monthly Revenue Trend")

plt.show()
```

## Key Business Insights

1. Total revenue exceeded $5.1 million.

2. Electronics generated the highest revenue ($1.83M), making it the most valuable product category.

3. Clothing was the second strongest category with over $1.53M in revenue.

4. The West region produced the highest sales revenue.

5. Revenue distribution across regions is relatively balanced, reducing geographic risk.
